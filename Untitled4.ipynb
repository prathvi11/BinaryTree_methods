```python
from langchain.document_loaders import PyPDFLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter

def load_and_split_pdf(pdf_path):
    loader = PyPDFLoader(pdf_path)
    documents = loader.load()

    splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
    chunks = splitter.split_documents(documents)

    return chunks

docs = load_and_split_pdf(pdf_path)
print(f"✅ Loaded and split into {len(docs)} chunks.")
```

```python
from langchain.embeddings import HuggingFaceEmbeddings

# Load MiniLM embeddings model
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("✅ HuggingFace Embeddings ready!")
```

```python
from langchain.vectorstores import FAISS

vectorstore = FAISS.from_documents(docs, embeddings)
retriever = vectorstore.as_retriever()
print("✅ Chunks embedded and stored in FAISS!")
```

```python
# Function to ask a question to the retriever
def ask_question(question):
    # Get the top 1 document based on the question
    result = retriever.get_relevant_documents(question)

    # Check if there's a valid result
    if result:
        return result[0].page_content  # Return the content of the most relevant chunk
    else:
        return "Sorry, no relevant information found."

# Example usage
response = ask_question("What is this PDF about?")
print("Answer:", response)
```

```python
# Function to keep asking questions until user decides to stop
def continuous_qa():
    print("💬 Ask me anything from the PDF! Type 'exit' to stop.")

    while True:
        question = input("Your Question: ")

        if question.lower() == 'exit':
            print("❌ Exiting... Goodbye!")
            break

        answer = ask_question(question)  # Using the earlier function to fetch the answer
        print("Answer:", answer)

# Start continuous Q&A
continuous_qa()
```
